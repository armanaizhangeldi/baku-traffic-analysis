```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
df=pd.read_csv('C:/Users/User/Desktop/python/traffic.csv')
df
```

```python
order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']

# Create a categorical type with the specified order
df['day'] = pd.Categorical(df['Day of the week'], categories=order, ordered=True)

# Sort the DataFrame by the 'day' column
sorted_df = df.sort_values('Day of the week')

sorted_df
```

```python
morning= df[df['Time'] == '7:00:00 AM']
morning_df=pd.DataFrame(morning)
morning_df
```

```python
midday= df[df['Time'] == '12:00:00 PM']
midday_df=pd.DataFrame(midday)
midday_df
```

```python
evening= df[df['Time'] == '5:00:00 PM']

evening_df=pd.DataFrame(evening)
evening_df
```

```python
midnight= df[df['Time'] == '12:00:00 AM']
midnight_df=pd.DataFrame(midnight)
midnight_df
```

```python
plt.plot(morning_df['Day of the week'], morning_df['Total'], color='cyan', label='Morning')
plt.plot(midday_df['Day of the week'], midday_df['Total'], color='green', label='Midday')
plt.plot(evening_df['Day of the week'], evening_df['Total'], color='orange', label='Evening')
plt.plot(midnight_df['Day of the week'], midnight_df['Total'], color='blue', label='Midnight')

# Set the labels for the axes
plt.xlabel('Days of the week')
plt.ylabel('Vehicle amount')

# Add the legend
plt.legend(ncol=2)  # Optionally add a title to the legend

# Show the plot
plt.show()
```

```python
total1 = df['CarCount'].mean()
total2 = df['BikeCount'].mean()
total3 = df['BusCount'].mean()
total4 = df['TruckCount'].mean()

# Labels and values for the pie chart
my_labels = ['CarCount', 'BikeCount', 'BusCount', 'TruckCount']
values = [total1, total2, total3, total4]

# Create the pie chart
plt.pie(values, labels=my_labels, autopct='%1.1f%%')
plt.title('Vehicles')
plt.show()
```

```python
oneday= df[df['Date'] == 7]
oneday_df=pd.DataFrame(oneday)
oneday_df = oneday_df.reset_index()
oneday_df
```

```python
x =oneday_df['Time']
y1 =oneday_df['CarCount']
y2 =oneday_df['BikeCount']
y3 =oneday_df['BusCount']
y4 =oneday_df['TruckCount']

plt.stackplot(x, y1, y2, y3, y4, labels = ['CarCount', 'BikeCount', 'BusCount', 'TruckCount'])
plt.xticks(size = 5.5, rotation = 90)
plt.legend(loc = 'upper right')
plt.show
```
